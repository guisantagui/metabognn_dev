```python
# --------------------------------------------------------------------------------------
# GEM layer functions
# --------------------------------------------------------------------------------------
def base_id(met_id):
    return met_id.rsplit("_", 1)[0]


def build_bipartite_graph(gem, excluded_mets, bookkeeping):
    G = nx.DiGraph()

    for rxn in gem.reactions:
        met_names = [m.name for m in rxn.metabolites]
        if any(name in bookkeeping for name in met_names):
            continue  # whole reaction dropped, as you're already doing

        has_genes = len(rxn.genes) > 0
        G.add_node(rxn.id, node_type="reaction", constitutive=not has_genes)

        reversible = rxn.lower_bound < 0 and rxn.upper_bound > 0

        for met in rxn.reactants:
            if base_id(met.name) in excluded_mets:
                continue
            G.add_node(met.id, node_type="metabolite")
            G.add_edge(met.id, rxn.id, edge_type="consumes")
            if reversible:
                # as a product in the reverse direction
                G.add_edge(rxn.id, met.id, edge_type="produces_rev")

        for met in rxn.products:
            if base_id(met.name) in excluded_mets:
                continue
            G.add_node(met.id, node_type="metabolite")
            G.add_edge(rxn.id, met.id, edge_type="produces")
            if reversible:
                G.add_edge(met.id, rxn.id, edge_type="consumes_rev")

    return G

def met_base_and_compartment(met):
    """Compartment-free metabolite id: 'atp_c' -> 'atp', Human-GEM 'MAM01371c' -> 'MAM01371'."""
    comp = met.compartment or ""
    if comp and met.id.endswith("_" + comp):
        return met.id[: -len(comp) - 1], comp
    if comp and met.id.endswith("[" + comp + "]"):
        return met.id[: -len(comp) - 2], comp
    if comp and met.id.startswith("MAM") and met.id.endswith(comp):
        return met.id[: -len(comp)], comp
    return base_id(met.id), comp


def read_lines(path):
    if path is None:
        return set()
    with open(path) as fh:
        return {l.strip() for l in fh if l.strip()}

def gem_gene_symbols(gem):
    sym = {}
    for g in gem.genes:
        sym[g.id] = g.annotation["hgnc.symbol"]
    return sym

def rule_to_symbols(rule, gene_sym):
    """Rewrite a GPR rule string with gene symbols so activity can be evaluated later."""
    if not rule:
        return ""
    return re.sub(r"[^\s()]+",
                  lambda m: m.group(0) if m.group(0).lower() in ("and", "or") else gene_sym.get(m.group(0), m.group(0)),
                  rule)

def build_gem_layer(gem, excluded_mets, bookkeeping, gene_sym):
    G = build_bipartite_graph(gem, excluded_mets, bookkeeping)
    nodes, edges = [], []
    for n, data in G.nodes(data=True):
        if data["node_type"] == "reaction":
            rxn = gem.reactions.get_by_id(n)
            nodes.append(dict(node_id=n, node_type="reaction", symbol="", base_met="", compartment="",
                              gpr=rule_to_symbols(rxn.gene_reaction_rule, gene_sym),
                              subsystem=str(rxn.subsystem or ""), constitutive=int(data["constitutive"])))
        else:
            met = gem.metabolites.get_by_id(n)
            b, c = met_base_and_compartment(met)
            nodes.append(dict(node_id=n, node_type="metabolite", symbol=met.name, base_met=b, compartment=c,
                              gpr="", subsystem="", constitutive=0))
    for s, d in G.edges():
        edges.append(dict(src=s, dst=d, relation="metabolic", directed=1, sign=0,
                          n_resources=1, n_references=0, resources="GEM"))
    # catalysis: every gene in the GPR points to its reaction (AND/OR kept in the rule string)
    for rxn_id in [n["node_id"] for n in nodes if n["node_type"] == "reaction"]:
        for g in gem.reactions.get_by_id(rxn_id).genes:
            edges.append(dict(src=gene_sym[g.id], dst=rxn_id, relation="catalysis", directed=1, sign=1,
                              n_resources=1, n_references=0, resources="GEM"))
    return pd.DataFrame(nodes), pd.DataFrame(edges)

def metabolite_index(gem, met_nodes):
    """Map external metabolite identifiers -> list of compartmentalised metabolite node ids."""
    present = set(met_nodes)
    idx = defaultdict(set)

    def norm(x):
        x = str(x).strip().upper()
        for p in ("CHEBI:", "HMDB:", "KEGG:", "MNXM:", "METANETX.CHEMICAL:"):
            if x.startswith(p) and p != "CHEBI:":
                x = x[len(p):]
        return x

    for met in gem.metabolites:
        if met.id not in present:
            continue
        b, _ = met_base_and_compartment(met)
        keys = {met.id, b, met.name}
        for k, v in (met.annotation or {}).items():
            for vv in (v if isinstance(v, list) else [v]):
                keys.add(vv)
                if k == "chebi" and not str(vv).upper().startswith("CHEBI:"):
                    keys.add("CHEBI:" + str(vv))
        for k in keys:
            if k:
                idx[norm(k)].add(met.id)
    return lambda q: sorted(idx.get(norm(q), ()))
```

```python
def count_tokens(x):
    if not isinstance(x, str) or not x:
        return 0
    return len({t.split(":")[-1] for t in re.split(r"[;,|]", x) if t})

def load_omnipath(path):
    df = pd.read_csv(path, sep="\t", dtype=str, keep_default_na=False)
    def flag(col, fallback=None):
        if col in df:
            return df[col].isin(["1", "True", "true"]).astype(int)
        return flag(fallback) if fallback else pd.Series(0, index=df.index)
    src = df["source_genesymbol"].to_list()
    dst = df["target_genesymbol"].to_list()
    directed = flag("consensus_direction", "is_directed")
    stim, inh = flag("consensus_stimulation", "is_stimulation"), flag("consensus_inhibition", "is_inhibition")
    sign = stim - inh
    n_res = df["sources"].map(count_tokens) if "sources" in df else pd.to_numeric(df.get("n_sources", 1))
    n_ref = df["references"].map(count_tokens) if "references" in df else pd.to_numeric(df.get("n_references", 0))
    typ = df["type"] if "type" in df else pd.Series("post_translational", index=df.index)
    # If its directed label as signalling, otherwise as PPI
    rel = np.where(
        typ.str.contains("transcriptional") & ~typ.str.contains("post"), "grn",
        np.where(directed == 1, "signaling", "ppi")
    )
    rel = np.where(typ == "small_molecule_protein", "met_regulation", rel)
    out = pd.DataFrame(
        dict(
            src=src, dst=dst, relation=rel, directed=directed, sign=sign,
            n_resources=n_res, n_references=n_ref,
            resources=df["sources"] if "sources" in df else "OmniPath"
        )
    )
    return out
```

```python
def load_trrust(path):
    df = pd.read_csv(path, sep="\t", header=None, dtype=str, names=["src", "dst", "mode", "pmids"])
    sign = df["mode"].map({"Activation": 1, "Repression": -1}).fillna(0).astype(int)
    return pd.DataFrame(dict(src=df.src, dst=df.dst, relation="grn", directed=1, sign=sign, n_resources=1,
                             n_references=df.pmids.map(count_tokens), resources="TRRUST"))
```

```python
def merge_edges(raw):
    """Collapse duplicates within a relation; undirected pairs are stored once (sorted)."""
    raw = raw[(raw.src != raw.dst) & (raw.src != "") & (raw.dst != "")].copy()
    und = raw.directed == 0
    a, b = raw.loc[und, "src"].values, raw.loc[und, "dst"].values
    raw.loc[und, "src"], raw.loc[und, "dst"] = np.minimum(a, b), np.maximum(a, b)
    raw["sign"] = pd.to_numeric(raw["sign"]).astype(int)
    g = raw.groupby(["relation", "src", "dst", "directed"], sort=False)
    merged = g.agg(sign_sum=("sign", "sum"), sign_abs=("sign", lambda s: s.abs().sum()),
                   n_res_max=("n_resources", "max"), n_references=("n_references", "max"),
                   resources=("resources", lambda s: ";".join(sorted({t for x in s for t in str(x).split(";") if t}))))
    merged = merged.reset_index()
    # conflicting signs (both + and - reported) -> 0 = unknown
    merged["sign"] = np.where(merged.sign_abs == merged.sign_sum.abs(), np.sign(merged.sign_sum), 0).astype(int)
    merged["n_resources"] = np.maximum(merged.n_res_max, merged.resources.map(count_tokens))
    return merged.drop(columns=["sign_sum", "sign_abs", "n_res_max"])
```

```python
def symb2ensembl(symbol_list, hgnc):
    symbol_to_ensembl = (
        hgnc.groupby("symbol")["ensembl_gene_id"]
        .apply(lambda x: "|".join(x.dropna().astype(str)))
        .to_dict()
    )
    alias_to_ensembl = defaultdict(list)
    for _, row in hgnc.dropna(subset=["alias_symbol"]).iterrows():
        for alias in row["alias_symbol"].split("|"):
            alias_to_ensembl[alias].append(str(row["ensembl_gene_id"]))
    alias_to_ensembl = {
        k: "|".join(sorted(set(v)))
        for k, v in alias_to_ensembl.items()
    }
    ensembl_list = []
    for gene in symbol_list:
        if gene in symbol_to_ensembl:
            ensembl_list.append(symbol_to_ensembl[gene])
        else:
            ensembl_list.append(alias_to_ensembl.get(gene, ""))
    return ensembl_list
```

```python
def load_chea(path):
    df = pd.read_csv(path, sep = '\t')
    src = df["source"].to_list()[1:df.shape[0]]
    dst = df["target"].to_list()[1:df.shape[0]]
    return pd.DataFrame(
        dict(
            src=src, dst=dst, relation="grn", directed=1,
            sign=0, n_resources=1, n_references=1, resources="ChEA"
        )
    )
```

```python
string_min_score = 700
```

```python
STRING_CHANNELS = {"experiments": ("experiments", "experimental"), "database": ("database",),
                   "textmining": ("textmining",), "coexpression": ("coexpression",)}
STRING_PRIOR = 0.041
```

```python
def load_string(links_path, info_path, channels, min_score):
    """Human STRING (e.g. 9606.protein.physical.links.detailed.v12.0.txt.gz) as undirected PPI.

    The score is recombined from the selected channels only, with STRING's formula
    (remove the prior from each channel, combine as 1 - prod(1 - s), add the prior back), so text mining
    and co-expression do not leak in through combined_score. Each contributing channel counts as one
    line of evidence (resources STRING_experiments;STRING_database)."""
    info = pd.read_csv(info_path, sep="\t", usecols=[0, 1], dtype=str)
    ensp2sym = dict(zip(info.iloc[:, 0], info.iloc[:, 1]))
    header = pd.read_csv(links_path, sep=" ", nrows=0).columns
    cols = {}
    for ch in channels:
        found = [c for c in STRING_CHANNELS.get(ch, (ch,)) if c in header]
        if found:
            cols[ch] = found[0]
    if not cols:
        print("[STRING] no channel columns found (non-detailed file?): using combined_score, which includes "
              "text mining and co-expression")
    out, n_raw = [], 0
    for df in pd.read_csv(links_path, sep=" ", chunksize=2_000_000):
        n_raw += len(df)
        df = df[df.protein1 < df.protein2]  # each pair is listed in both orientations
        if cols:
            s = np.stack([df[c].values / 1000.0 for c in cols.values()], 1)
            s_nop = np.clip((s - STRING_PRIOR) / (1 - STRING_PRIOR), 0, None)
            comb = 1 - np.prod(1 - s_nop, 1)
            score = comb + STRING_PRIOR * (1 - comb)
            support = s_nop > 0
        else:
            score = df.combined_score.values / 1000.0
            support = np.ones((len(df), 1), bool)
        keep = score >= min_score / 1000.0
        names = np.array([f"STRING_{ch}" for ch in cols] or ["STRING"])
        res = [";".join(names[row]) for row in support[keep]]
        out.append(pd.DataFrame(dict(src=df.protein1.values[keep], dst=df.protein2.values[keep], relation="ppi",
                                     directed=0, sign=0, n_resources=[r.count(";") + 1 for r in res],
                                     n_references=0, resources=res)))
    out = pd.concat(out, ignore_index=True)
    out["src"], out["dst"] = out.src.map(ensp2sym), out.dst.map(ensp2sym)
    out = out.dropna(subset=["src", "dst"])
    print(f"[STRING] {n_raw // 2} pairs, {len(out)} with score >= {min_score} on {list(cols) or ['combined']}")
    return out
```
