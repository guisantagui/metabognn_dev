# Building the coupled metabolic–signaling network, step by step

This notebook runs the network build of `build_network.py` one step at a time, so every intermediate table can be
inspected. It calls the **same functions** as the script (`import build_network as bn`), so what you verify here is
what the script produces.

**Why the network looks the way it does.** The metabolic signatures (MetChange, METAflux) are computed from
expression through Human-GEM. A drug can therefore only change them by changing the *transcription of metabolic
enzymes*:

```
drug target ──signaling / modification──▶ TF ──regulon──▶ enzyme gene ──catalysis──▶ reaction ◀─▶ metabolite
```

Each input resource provides one piece of that route:

| relation | edges | source in `data/networks` |
|---|---|---|
| `metabolic` | metabolite ↔ reaction | Human-GEM 2.0 |
| `catalysis` | enzyme gene → reaction | Human-GEM gene rules |
| `grn` | TF → gene, signed | TRRUST (+ CollecTRI when downloaded) |
| `grn_binding` | TF → gene, unsigned ChIP binding | ChEA (Harmonizome) |
| `signaling` | protein → protein, directed | OmniPath core |
| `enzyme_substrate` | kinase/modifier → substrate | OmniPath enz_sub |
| `ppi` | undirected | STRING physical, HIPPIE, undirected OmniPath rows |

Sections: **1** identifiers · **2** GEM layer · **3** each protein-layer source · **4** identifier mapping ·
**5** merging duplicates · **6** curation · **7** assembly and checks · **8** coverage of the transcriptional route ·
**9** outputs.

## 1. Identifiers: one namespace for all proteins

Every resource names proteins differently: OmniPath and TRRUST use gene symbols (some outdated, some UniProt
accessions when no symbol exists), HIPPIE and ChEA use Entrez ids, Human-GEM uses Ensembl gene ids, STRING its own
protein names. All of them are mapped to **HGNC approved symbols** of protein-coding genes (plus complex-locus
constituents, where HGNC files the UGT1A glucuronidation enzymes):

1. an approved symbol maps to itself;
2. Entrez, UniProt and Ensembl ids, previous symbols and aliases map to the approved symbol **only when they point
   to a single gene** (an ambiguous alias is skipped rather than guessed);
3. an alias never overrides an approved symbol (e.g. if `X` is approved for one gene and an alias of another,
   `X` stays the approved gene).

Edges whose endpoint cannot be mapped are dropped in section 4 (and counted).

````python
hgnc = bn.HGNC(paths["hgnc"])
print(f"approved protein-coding symbols: {len(hgnc.approved):,}")
print(f"identifiers that map to a symbol: {len(hgnc.map):,}")
print(f"ambiguous identifiers skipped:   {len(hgnc.ambiguous):,}")
pd.DataFrame({"query": ["TP53", "7157", "P04637", "ENSG00000141510", "p53", "MYC"],
              "maps_to": [hgnc(q) for q in ["TP53", "7157", "P04637", "ENSG00000141510", "p53", "MYC"]]})
````

## 2. GEM layer (Human-GEM 2.0)

### 2.1 Load the model

````python
gem = bn.load_gem(paths["gem"])
print(f"{len(gem.reactions):,} reactions, {len(gem.metabolites):,} metabolites, {len(gem.genes):,} genes")
````

### 2.2 Metabolites to exclude

Four groups of metabolites are removed **as nodes** (the reaction stays, it just loses that participant):

* **currency** – cofactors and small inorganic species (ATP, NAD(P)H, H₂O, H⁺ …). They take part in thousands of
  reactions; kept as nodes they would connect almost every reaction to every other in two hops, so propagation
  would spread everywhere and carry no information.
* **bookkeeping** – biomass components, protein/ACP carriers, lipoproteins. Additionally, **any reaction containing
  one of these is dropped entirely**: they are modelling devices, not metabolism.
* **macromolecular** – tRNAs, DNA, RNA.
* **pool** – lipid "pool" pseudo-metabolites (name contains `pool`).

First check: every listed name must exist in the model. A name that matches nothing (typo, different naming in
Human-GEM) silently excludes nothing.

````python
excluded, bookkeeping = bn.gem_exclusions(gem)
names = pd.Series([m.name for m in gem.metabolites]).value_counts()
check = pd.DataFrame([(n, grp, int(names.get(n, 0)))
                      for grp, s in [("currency", bn.CURRENCY), ("bookkeeping", bn.BOOKKEEPING),
                                     ("macromolecular", bn.MACROMOLECULAR)] for n in sorted(s)],
                     columns=["name", "group", "entries_in_model"])
print(f"{len(excluded):,} excluded names in total "
      f"({len(excluded - bn.CURRENCY - bn.BOOKKEEPING - bn.MACROMOLECULAR):,} 'pool' metabolites)")
print("names that match no metabolite in Human-GEM:")
check[check.entries_in_model == 0]
````

### 2.3 Change from `build_gem_layer.ipynb`: thioredoxin is currency, not bookkeeping

In the notebook, `thioredoxin` and `mitothioredoxin` were in both the currency and the bookkeeping sets. As
bookkeeping they make **the whole reaction disappear**. Thioredoxin is an electron carrier like NADPH, so it should
only be removed as a node. The reactions that the old setting would have dropped:

````python
thx = [r for r in gem.reactions if any(m.name in ("thioredoxin", "mitothioredoxin") for m in r.metabolites)]
pd.DataFrame([(r.id, r.subsystem, r.name, r.gene_reaction_rule[:60]) for r in thx],
             columns=["reaction", "subsystem", "name", "gene rule"]).sort_values("subsystem")
````

These include ribonucleotide reductase (dNTP synthesis, nucleotide metabolism), which is relevant for A549
proliferation. To restore the old behaviour, add the two names back to `bn.BOOKKEEPING`.

### 2.4 Change from `build_gem_layer.ipynb`: exclusions match full names

The notebook tested `base_id(met.name) in excluded_mets`, where `base_id` cuts at the last `_` (meant for ids such
as `atp_c`). Human-GEM ids carry no `_`, but some *names* do, and these were truncated before matching:

````python
with_us = sorted({m.name for m in gem.metabolites if "_" in m.name})
base_id = lambda x: x.rsplit("_", 1)[0]  # as in build_gem_layer.ipynb
pd.DataFrame({"name": with_us, "base_id(name)": [base_id(n) for n in with_us],
              "truncated name was excluded": [base_id(n) in excluded for n in with_us],
              "full name is excluded": [n in excluded for n in with_us]})
````

### 2.5 Build the metabolite ↔ reaction graph

Edges follow reaction direction: substrate → reaction → product. A **reversible** reaction (lower bound < 0 and
upper bound > 0) also gets the reverse edges, so it can be traversed both ways. A reaction without gene rules is
flagged `constitutive` (spontaneous or transport without an annotated transporter); the prediction pipeline gives
it a default activity.

````python
gene_sym = bn.gem_gene_symbols(gem, hgnc)
nodes_gem, edges_gem, G = bn.build_gem_layer(gem, excluded, bookkeeping, gene_sym)

rxn = nodes_gem[nodes_gem.node_type == "reaction"]
dropped = [r.id for r in gem.reactions if r.id not in G]
isolated = [n for n in rxn.node_id if G.degree(n) == 0]
reversible = sum(1 for r in gem.reactions if r.id in G and r.lower_bound < 0 < r.upper_bound)
pd.Series({
    "reactions in model": len(gem.reactions),
    "reactions dropped (bookkeeping)": len(dropped),
    "reactions kept": len(rxn),
    "  of which reversible": reversible,
    "  of which constitutive (no genes)": int(rxn.constitutive.sum()),
    "  of which isolated (all metabolites excluded)": len(isolated),
    "metabolite nodes (compartment-specific)": int((nodes_gem.node_type == "metabolite").sum()),
    "distinct metabolites (compartment-free)": nodes_gem.loc[nodes_gem.node_type == "metabolite", "base_met"].nunique(),
    "metabolic edges": int((edges_gem.relation == "metabolic").sum()),
    "catalysis edges": int((edges_gem.relation == "catalysis").sum()),
}).to_frame("count")
````

Subsystems of the reactions dropped as bookkeeping (check that nothing you need is lost):

````python
pd.Series([gem.reactions.get_by_id(r).subsystem for r in dropped]).value_counts().head(15).to_frame("reactions dropped")
````

### 2.6 Remaining metabolite hubs

After the exclusions, the most connected metabolites should be genuine pathway intermediates. Any remaining
cofactor-like species near the top (high degree in many compartments) is a candidate for the currency list:

````python
met = nodes_gem[nodes_gem.node_type == "metabolite"].set_index("node_id")
deg = pd.Series({n: G.degree(n) for n in met.index})
hubs = (met.assign(degree=deg).groupby("symbol").agg(degree=("degree", "sum"), compartments=("compartment", "nunique"))
        .sort_values("degree", ascending=False))
hubs.head(25)
````

````python
fig, ax = plt.subplots(figsize=(6, 3.5))
ax.hist(deg, bins=np.logspace(0, np.log10(deg.max() + 1), 40))
ax.set(xscale="log", yscale="log", xlabel="edges per metabolite node", ylabel="metabolite nodes",
       title="Metabolite degree after exclusions")
plt.show()
````

### 2.7 Genes: Ensembl → HGNC symbol

Human-GEM gene rules use Ensembl gene ids. They are mapped to symbols through HGNC; when HGNC lists a different
Ensembl id for the gene (it happens for SOD2 and SLC37A4), the model's own `hgnc.symbol` annotation is mapped
instead. The gene rule is stored twice on each reaction node: with symbols (`gpr`) and with the
original Ensembl ids (`gpr_ensembl`), so expression in either namespace can be used to compute reaction activity
(AND = complex, minimum; OR = isozymes, maximum).

````python
genes = pd.DataFrame([(g.id, g.annotation.get("hgnc.symbol", ""), hgnc(g.id), gene_sym[g.id]) for g in gem.genes],
                     columns=["ensembl", "model_symbol", "via_ensembl", "symbol"])
ok = genes.symbol.isin(hgnc.approved)
print(f"{ok.sum():,}/{len(genes):,} GEM genes have an HGNC protein-coding symbol")
print("mapped through the model symbol because HGNC lists another Ensembl id:")
display(genes[ok & (genes.via_ensembl == "")])
print("updated symbols (model annotation is outdated), e.g.:")
display(genes[ok & (genes.symbol != genes.model_symbol)].head(5))
print("not mapped (their catalysis edges are dropped in section 4):")
genes[~ok]
````

## 3. Protein-layer sources

Each loader returns the same columns: `src, dst, relation, directed, sign, resources, pmids, n_resources,
n_references, origins`. Identifiers are still raw here; they are mapped in section 4.

Two evidence measures are carried for every edge and used for curation in section 6:

* **resources** – the databases that report the edge. For OmniPath, derived resources are folded into their
  parent (`PhosphoSite_ProtMapper` → `PhosphoSite`), otherwise one curation effort counts several times.
* **pmids** – the publications supporting it. Unlike resource counts, PMIDs do not double-count when several
  aggregators (OmniPath, HIPPIE, STRING) re-distribute the same BioGRID/IntAct experiment.

### 3.1 OmniPath core interactions

````python
op_raw = bn.read_table(paths["omnipath"], index_col=0)
print(op_raw.shape)
op_raw.head(3)
````

**Complexes.** More than half of the rows have a protein complex as source or target (mostly SIGNOR), written as
`COMPLEX:<uniprot>_<uniprot>…` with gene symbols joined by `_`. A complex is not a gene, so it would be lost when

*subunit → target* per subunit. Example:

````python
ex = op_raw[op_raw.source.str.startswith("COMPLEX:")].head(1)
display(ex[["source", "source_genesymbol", "target_genesymbol", "sources"]])
bn.expand_complexes(ex.copy())[["source_genesymbol", "target_genesymbol"]]
````

**Direction and sign.** `consensus_direction` is True when the resources agree on the direction; those rows become
`signaling` edges with the consensus sign (+1 stimulation, −1 inhibition, 0 unknown or both). Rows without a
consensus direction are kept as undirected `ppi`. All rows without consensus direction are also unsigned:

````python
pd.crosstab(op_raw.consensus_direction,
            op_raw.consensus_stimulation + "/" + op_raw.consensus_inhibition,
            margins=True).rename_axis(columns="stimulation/inhibition")
````

### 3.5 STRING physical links – rescoring

STRING's `combined_score` mixes evidence channels. The physical file has three: `experimental`, `database` and
`textmining`. Text mining means two genes are mentioned together, not that they bind. The score is therefore
recomputed from **experimental + database only**, with STRING's own formula: remove the prior (0.041) from each
channel, combine as 1 − ∏(1 − sᵢ), add the prior back. Each channel that supports a pair counts as one resource.

````python
sl = pd.read_csv(paths["string_links"], sep=" ")
sl = sl[sl.protein1 < sl.protein2]
ours, support = bn.string_score(np.stack([sl.experimental / 1000, sl.database / 1000], 1))
thr = cfg.string_min_score / 1000
pd.Series({
    "pairs": len(sl),
    f"combined_score >= {cfg.string_min_score}": int((sl.combined_score / 1000 >= thr).sum()),
    f"experiments+database score >= {cfg.string_min_score}": int((ours >= thr).sum()),
    "  passing only because of text mining": int(((sl.combined_score / 1000 >= thr) & (ours < thr)).sum()),
}).to_frame("count")
````

````python
fig, ax = plt.subplots(figsize=(4.5, 4.5))
idx = np.random.default_rng(0).choice(len(sl), 20000, replace=False)
ax.scatter(sl.combined_score.to_numpy()[idx] / 1000, ours[idx], s=2, alpha=0.3)
ax.axhline(thr, color="grey", lw=0.8); ax.axvline(thr, color="grey", lw=0.8)
ax.set(xlabel="STRING combined_score", ylabel="experiments + database score", title="STRING rescoring (20k pairs)")
plt.show()
````

````python
st = bn.load_string(paths["string_links"], paths["string_info"], cfg.string_channels.split(","), cfg.string_min_score)
st.resources.value_counts().to_frame("pairs by supporting channels")
````

### 3.7 CollecTRI

CollecTRI is a curated, signed TF regulon collection that integrates a dozen resources (ExTRI, TRRUST, NTNU,
GEREDB, TFactS, HTRI, DoRothEA-A, SIGNOR …). It is downloaded by `omnipath_dl.py` and loaded as `grn` edges.

* Each row lists the original resources with a `_CollecTRI` suffix (`TRRUST_CollecTRI` → `TRRUST`, so it merges
  with the TRRUST file) plus the aggregator `CollecTRI`, which is dropped so it does not add a resource to every edge.
* TF complexes (AP-1, NF-κB) are written as `COMPLEX:` and expanded into subunits, as for OmniPath.
* Every edge is signed. CollecTRI assigns activation when the literature does not state a mode, so a `+1` from
  CollecTRI is weaker evidence than a `−1`.

````python
ct_raw = bn.read_table(paths["collectri"], index_col=0)
print(f"{len(ct_raw):,} rows, {ct_raw.source.str.startswith('COMPLEX:').sum():,} with a TF complex")
ct_raw.sources.str.split(";").explode().str.replace("_CollecTRI", "").value_counts().head(12).to_frame("rows")
````

## 5. Merge duplicate edges

The same edge is often reported by several sources (e.g. OmniPath and HIPPIE both carry BioGRID interactions;
expanded complexes produce repeated subunit edges). Within each relation, duplicates are collapsed:

* undirected edges are stored once, with the endpoints sorted;
* resources, PMIDs and origins are unioned, and the counts are recomputed from the unions;
* sign: if one source says activation and another inhibition, the merged sign is 0 (unknown).

Note that duplicates are merged **within** a relation: the same pair can appear both as `signaling` and as `ppi`
(handled in section 6).

````python
merged = bn.merge_edges(mapped)
pd.DataFrame({"rows before": mapped.groupby("relation").size(),
              "edges after": merged.groupby("relation").size()})
````

Edges with the most supporting resources (sanity check: these should be well-known interactions):

````python
merged.sort_values("n_resources", ascending=False).head(8)[["src", "dst", "relation", "sign", "n_resources",
                                                             "n_references", "origins"]]
````

Edges where sources disagree on the sign (set to 0):

````python
signs = mapped[mapped.sign != 0].groupby(bn.KEYS).sign.agg(lambda s: s.nunique() > 1)
conflict = signs[signs].reset_index()[bn.KEYS]
print(f"{len(conflict)} edges with conflicting signs")
conflict.head()
````

## 6. Curation

| relation | kept if |
|---|---|
| `signaling`, `grn` | ≥ `min_directed_resources` resources **or** ≥ `min_directed_refs` PMIDs |
| `enzyme_substrate` | ≥ `enzsub_min_primary` primary resources **or** ≥ `enzsub_min_refs` PMIDs, and substrate is a TF |
| `ppi` | ≥ `min_ppi_resources` resources **or** ≥ `min_ppi_refs` PMIDs |
| `grn_binding` | always (no per-edge evidence in Harmonizome ChEA) |
| `metabolic`, `catalysis` | always (GEM) |

Then a `ppi` edge is removed when the same pair already has a directed edge (`signaling`, `enzyme_substrate`,
`grn`, `grn_binding`): the directed edge carries strictly more information.

Finally each edge gets a weight from its evidence: 1 − 0.5ⁿ for n resources (1 → 0.5, 2 → 0.75, 3 → 0.875), and
1 for GEM edges. The pipeline mixes relations with its own `--alpha` weights on top of this.

````python
edges, log = bn.curate(merged, cfg)
display(pd.Series(log, dtype=object).to_frame("curation log"))
pd.DataFrame({"merged": merged.groupby("relation").size(), "curated": edges.groupby("relation").size()})
````

````python
fig, axes = plt.subplots(1, 2, figsize=(10, 3.5))
for ax, col in zip(axes, ["n_resources", "n_references"]):
    for r in ["signaling", "grn", "ppi", "enzyme_substrate"]:
        v = edges.loc[edges.relation == r, col].clip(upper=20)
        ax.hist(v, bins=np.arange(0, 22) - 0.5, histtype="step", label=r)
    ax.set(xlabel=f"{col} (clipped at 20)", ylabel="edges", yscale="log")
axes[0].legend()
plt.show()
````

## 7. Assemble nodes, prune, check

Protein nodes are every symbol used by a kept edge; each gets its Ensembl id from HGNC (the LINCS expression
matrix is in Ensembl ids). With `cfg.prune`, connected components that contain no metabolite are removed: a
protein that cannot reach the metabolic layer by any path can never receive signal.

````python
nodes = bn.assemble_nodes(nodes_gem, edges, hgnc)
n_before = nodes.node_type.value_counts()
if cfg.prune:
    nodes, edges = bn.keep_metabolic_components(nodes, edges)
pd.DataFrame({"before pruning": n_before, "after pruning": nodes.node_type.value_counts()})
````

Structural checks (unique nodes, all endpoints exist, no self-loops, no duplicate edges):

````python
bn.check(nodes, edges)
print("all checks passed")
````

How the protein layers overlap. **Enzymes** are catalysis sources, **regulated enzymes** are enzymes that are the
target of a `grn` edge (the entry point of the transcriptional route), **TFs** are `grn` sources:

````python
by_rel = lambda r, col: set(edges.loc[edges.relation == r, col])
enzymes = by_rel("catalysis", "src")
sets = {
    "enzymes (GEM)": enzymes,
    "regulated enzymes (grn target)": enzymes & by_rel("grn", "dst"),
    "regulated enzymes (incl. ChEA)": enzymes & (by_rel("grn", "dst") | by_rel("grn_binding", "dst")),
    "TFs (grn)": by_rel("grn", "src"),
    "TFs (ChEA)": by_rel("grn_binding", "src"),
    "proteins in signaling": by_rel("signaling", "src") | by_rel("signaling", "dst"),
    "TFs that are signaling targets": by_rel("grn", "src") & by_rel("signaling", "dst"),
}
pd.Series({k: len(v) for k, v in sets.items()}).to_frame("proteins")
````

### 7.1 How much does ChEA overlap with the other layers?

ChEA is ChIP binding, so its TF → gene pairs are compared with the curated regulons (`grn`: TRRUST, CollecTRI) for
the TFs both cover. The **expected** overlap is what random target lists of the same sizes would share
(Σ over TFs of |ChEA targets| × |regulon targets| / number of genes); the ratio observed / expected says whether
ChEA binding agrees with curated regulation more than chance.

````python
pairs = lambda df: set(zip(df.src, df.dst))
chea_e = edges[edges.relation == "grn_binding"]
n_genes = int((nodes.node_type == "protein").sum())
rows = []
for name, grn_e in [("grn (all)", edges[edges.relation == "grn"]),
                    ("TRRUST", edges[(edges.relation == "grn") & edges.origins.str.contains("TRRUST")]),
                    ("CollecTRI", edges[(edges.relation == "grn") & edges.origins.str.contains("CollecTRI")])]:
    shared_tfs = set(chea_e.src) & set(grn_e.src)
    c = chea_e[chea_e.src.isin(shared_tfs)]
    g = grn_e[grn_e.src.isin(shared_tfs)]
    both = pairs(c) & pairs(g)
    expected = sum(len(ct) * len(gt) / n_genes for (_, ct), (_, gt) in
                   zip(sorted(c.groupby("src").dst), sorted(g.groupby("src").dst)))
    rows.append(dict(regulons=name, shared_TFs=len(shared_tfs), ChEA_pairs=len(c), regulon_pairs=len(g),
                     overlap=len(both), pct_of_regulon_in_ChEA=round(100 * len(both) / len(g), 1),
                     pct_of_ChEA_in_regulon=round(100 * len(both) / len(c), 2),
                     expected_by_chance=round(expected, 1), fold_over_chance=round(len(both) / expected, 2)))
pd.DataFrame(rows).set_index("regulons")
````

ChEA pairs that are also connected in another layer (any direction), and what ChEA adds that nothing else covers.
Compared with `merged` (before curation), because curation removes PPI edges whose pair has a directed edge,
ChEA included:

````python
undirected = lambda df: set(map(frozenset, zip(df.src, df.dst)))
chea_pairs = undirected(chea_e)
layers = {r: undirected(merged[merged.relation == r]) for r in ["grn", "signaling", "enzyme_substrate", "ppi"]}
summary = {f"ChEA pairs also in {r}": len(chea_pairs & s) for r, s in layers.items()}
covered = set().union(*layers.values())
summary["ChEA pairs found only in ChEA"] = len(chea_pairs - covered)
summary["ChEA TFs without any curated regulon (grn)"] = len(set(chea_e.src) - set(edges.src[edges.relation == "grn"]))
enz = set(edges.src[edges.relation == "catalysis"])
summary["GEM enzymes regulated in grn"] = len(enz & set(edges.dst[edges.relation == "grn"]))
summary["GEM enzymes with a ChEA TF but no grn TF"] = len((enz & set(chea_e.dst)) - set(edges.dst[edges.relation == "grn"]))
pd.Series(summary).to_frame("count")
````

## 8. Coverage of the transcriptional route

The key question for the model: **from how many proteins can a signal reach the metabolic signature through the
route the signature can see?** For every protein, compute the shortest directed path (over `signaling`,
`enzyme_substrate`, `grn`) to a TF that has a signed `grn` edge to a GEM enzyme. A drug whose targets are all
further than `cfg.max_hops` (the number of propagation steps in the pipeline) cannot be explained by the model.

````python
dist, tfs = bn.transcriptional_route(nodes, edges)
prot = (nodes.node_type == "protein").to_numpy()
d = pd.Series(dist[prot], index=nodes.node_id[prot])
print(f"{len(tfs)} TFs with signed regulon edges to GEM enzymes")
print(f"proteins within {cfg.max_hops} hops: {(d <= cfg.max_hops).mean():.3f}; unreachable: {np.isinf(d).mean():.3f}")
d.replace(np.inf, -1).astype(int).value_counts().sort_index().rename({-1: "unreachable"}).to_frame("proteins")
````

For comparison, the same distance when ChEA binding also counts as regulation (an upper bound on what the
unsigned layer could add):

````python
grn_like = edges.copy()
grn_like.loc[grn_like.relation == "grn_binding", "relation"] = "grn"
dist2, tfs2 = bn.transcriptional_route(nodes, grn_like)
print(f"{len(tfs2)} TFs; proteins within {cfg.max_hops} hops: {(dist2[prot] <= cfg.max_hops).mean():.3f}")
````

**Drug targets (optional).** Set `DRUG_TARGETS` to a TSV with columns `drug, target` to see how many of your drugs
have at least one target on the route, and which drugs are also GEM metabolites (those metabolite nodes should be
removed from the drug's own signatures).

````python
DRUG_TARGETS = None  # e.g. DATA / "drugs" / "drug_targets.tsv"

drug_targets = matches = None
if DRUG_TARGETS:
    drug_targets = bn.read_table(DRUG_TARGETS).iloc[:, :2]
    drug_targets.columns = ["drug", "target"]
    drug_targets["target"] = drug_targets.target.map(hgnc)
    matches = bn.drug_metabolite_matches(drug_targets, nodes)
    display(matches)
lines = bn.report(nodes, edges, cfg, drug_targets)
print("\n".join(lines))
````

## Open decisions

* **ChEA hubs** – some TFs bind thousands of genes (section 3.4). Keep the low pipeline weight, or cap targets per TF.
* **Metabolite hubs** – L-carnitine is already in the currency list; review the top of section 2.6 for others.
* **PPI** – the largest layer after ChEA; its pipeline weight (0.25) and `cfg.max_ppi_degree` are the levers.
* **enz_sub** – `cfg.enzsub_substrates = "tf"` keeps only modifications of TFs; `"all"` for an ablation.
